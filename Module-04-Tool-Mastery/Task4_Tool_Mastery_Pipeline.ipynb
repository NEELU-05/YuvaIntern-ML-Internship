## 1. Environment & Library Ingestion

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import classification_report, roc_auc_score, confusion_matrix

df = pd.read_csv("benchmark_modeling_data.csv")
print("Dataset Shape:", df.shape)
display(df.head())
```

## 2. Preprocessing Architecture: ColumnTransformer & Scikit-Learn Pipeline
We assemble a leakage-free preprocessing pipeline that scales continuous numeric variables and one-hot encodes categorical factors.

```python
X = df.drop(columns=["Churn"])
y = df["Churn"]

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)

numeric_features = ["Tenure", "MonthlyCharges", "TotalCharges"]
categorical_features = ["SeniorCitizen", "Contract", "InternetService", "TechSupport", "PaymentMethod"]

preprocessor = ColumnTransformer(transformers=[
    ("num", Pipeline([("imputer", SimpleImputer(strategy="median")), ("scaler", StandardScaler())]), numeric_features),
    ("cat", Pipeline([("imputer", SimpleImputer(strategy="most_frequent")), ("onehot", OneHotEncoder(drop="first", handle_unknown="ignore"))]), categorical_features)
])

print("Preprocessor successfully defined.")
```

## 3. Model Training & Cross-Validation Benchmarking

```python
models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=150, max_depth=8, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, learning_rate=0.1, random_state=42)
}

for name, model in models.items():
    pipe = Pipeline([("prep", preprocessor), ("clf", model)])
    scores = cross_val_score(pipe, X_train, y_train, cv=5, scoring="roc_auc")
    pipe.fit(X_train, y_train)
    y_prob = pipe.predict_proba(X_test)[:, 1]
    test_auc = roc_auc_score(y_test, y_prob)
    print(f"{name:20s} | 5-Fold CV AUC: {scores.mean():.4f} (±{scores.std():.4f}) | Test AUC: {test_auc:.4f}")
```

## 4. Key Takeaways & Portfolio Artifacts
- **Pipeline Encapsulation:** Zero data leakage achieved by embedding `StandardScaler` inside the cross-validation folds.
- **Model Performance:** Gradient Boosting and Random Forest achieve the strongest discrimination metrics (ROC-AUC > 0.85).
- **Top Predictors:** Tenure, TotalCharges, and Contract Type serve as the primary drivers of customer behavior.
