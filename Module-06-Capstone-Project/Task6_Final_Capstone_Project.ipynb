## 1. Dataset Ingestion & Feature Engineering

```python
import pandas as pd
import numpy as np
import joblib

df = pd.read_csv("churnguard_capstone_data.csv")
print("Capstone Dataset Shape:", df.shape)
display(df.head())
print("\nClass Distribution:")
print(df["Churn"].value_counts(normalize=True).round(3))
```

## 2. Model Training & 5-Fold Stratified Cross-Validation Benchmark

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, roc_auc_score

# Load serialized winning pipeline
pipeline = joblib.load("churnguard_model.joblib")
print("Loaded Pipeline Steps:", pipeline.named_steps.keys())

X = df.drop(columns=["CustomerID", "Churn"])
y = df["Churn"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)

y_pred = pipeline.predict(X_test)
y_prob = pipeline.predict_proba(X_test)[:, 1]

print("\nHoldout Evaluation Report:")
print(classification_report(y_test, y_pred))
print(f"Holdout ROC-AUC Score: {roc_auc_score(y_test, y_prob):.4f}")
```
