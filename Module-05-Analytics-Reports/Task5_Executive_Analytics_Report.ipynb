## 1. Data Ingestion & Financial Metric Computation

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("executive_retention_data.csv")
print("Account Count:", len(df))
display(df.head())

total_arr = df['MRR'].sum() * 12
at_risk_arr = df[df['RiskCategory'] == 'High Risk']['MRR'].sum() * 12
print(f"Total Active ARR : ${total_arr:,.2f}")
print(f"ARR at Churn Risk: ${at_risk_arr:,.2f} ({(at_risk_arr/total_arr)*100:.1f}%)")
```

## 2. Strategic 4-Quadrant Account Breakdown

```python
quad_summary = df.groupby("StrategicQuadrant").agg(
    Accounts=("AccountID", "count"),
    TotalMRR=("MRR", "sum"),
    AvgCLV=("CustomerLifetimeValue", "mean"),
    AvgRisk=("PredictedChurnRisk", "mean")
).round(2)

quad_summary["ARR Exposure"] = quad_summary["TotalMRR"] * 12
display(quad_summary)
```
