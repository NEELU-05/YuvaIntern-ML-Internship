## 1. Python Programming Fundamentals for Data Science
In this section, we demonstrate core Python syntax, primitive and compound data structures (lists, tuples, sets, dictionaries), list comprehensions, functional constructs, and object-oriented encapsulation for data pipelines.

```python
import math
import random
from typing import List, Dict, Tuple, Any, Optional

# 1. List manipulation & comprehensions
raw_sensor_readings = [23.5, 25.1, 22.8, 28.4, 24.0, 30.2, 21.9, 26.5]
min_val, max_val = min(raw_sensor_readings), max(raw_sensor_readings)

# Min-max feature scaling via list comprehension
scaled_readings = [round((x - min_val) / (max_val - min_val), 4) for x in raw_sensor_readings]
print(f"Raw Readings   : {raw_sensor_readings}")
print(f"Scaled [0, 1]  : {scaled_readings}")
```

### 1.2 Object-Oriented Pipeline Abstraction

```python
class DataPipelineProcessor:
    """Demonstrates OOP principles (encapsulation, abstraction) for clean data pipelines."""
    def __init__(self, data: List[Optional[float]], name: str = "TelemetryPipeline"):
        self._name = name
        self._raw_data = data
        self._cleaned_data: List[float] = []
        
    def clean_missing_and_negatives(self) -> "DataPipelineProcessor":
        self._cleaned_data = [x for x in self._raw_data if x is not None and x >= 0]
        return self
        
    def get_data(self) -> List[float]:
        return self._cleaned_data

stream = [14.2, 19.5, -3.0, None, 22.1, 18.7, 35.0, 16.4]
processor = DataPipelineProcessor(stream)
cleaned_data = processor.clean_missing_and_negatives().get_data()
print(f"Pipeline '{processor._name}' processed dataset: {cleaned_data}")
```

## 2. Statistical Foundations for Machine Learning
A machine learning model is only as sound as the statistical principles on which it is built. Below, we implement measures of central tendency (mean, median, mode), measures of dispersion (variance, standard deviation, interquartile range), and hypothesis testing.

```python
class StatisticalAnalyzer:
    def __init__(self, sample: List[float]):
        self.sample = sorted(sample)
        self.n = len(sample)

    def mean(self) -> float:
        return sum(self.sample) / self.n

    def median(self) -> float:
        mid = self.n // 2
        if self.n % 2 == 0:
            return (self.sample[mid - 1] + self.sample[mid]) / 2.0
        return float(self.sample[mid])

    def variance(self, ddof: int = 1) -> float:
        m = self.mean()
        return sum((x - m) ** 2 for x in self.sample) / (self.n - ddof)

    def std_dev(self) -> float:
        return math.sqrt(self.variance())

    def iqr_bounds(self) -> Tuple[float, float, float, Tuple[float, float]]:
        def percentile(p):
            k = (self.n - 1) * p
            f, c = math.floor(k), math.ceil(k)
            if f == c:
                return self.sample[int(k)]
            return self.sample[f] * (c - k) + self.sample[c] * (k - f)
        q1, q3 = percentile(0.25), percentile(0.75)
        iqr = q3 - q1
        return q1, q3, iqr, (q1 - 1.5 * iqr, q3 + 1.5 * iqr)

stats = StatisticalAnalyzer(cleaned_data)
q1, q3, iqr, (lb, ub) = stats.iqr_bounds()
print(f"N          : {stats.n}")
print(f"Mean       : {stats.mean():.3f}")
print(f"Median     : {stats.median():.3f}")
print(f"Std Dev    : {stats.std_dev():.3f}")
print(f"Q1, Q3, IQR: ({q1:.2f}, {q3:.2f}, {iqr:.2f})")
print(f"IQR Outlier Bounds: [{lb:.2f}, {ub:.2f}]")
```

### 2.2 Correlation & Hypothesis Testing (Two-Sample t-Test)

```python
def pearson_r(x: List[float], y: List[float]) -> float:
    n = len(x)
    mx, my = sum(x) / n, sum(y) / n
    num = sum((xi - mx) * (yi - my) for xi, yi in zip(x, y))
    den = math.sqrt(sum((xi - mx)**2 for xi in x) * sum((yi - my)**2 for yi in y))
    return num / den if den != 0 else 0.0

feature_a = [10, 12, 14, 16, 18, 20, 22, 24]
feature_b = [25, 29, 34, 38, 45, 51, 56, 62]
print(f"Pearson correlation coefficient: {pearson_r(feature_a, feature_b):.4f}")
```

## 3. Data Types & Encoding Strategies
Data in real-world machine learning tasks spans across multiple representations: **Quantitative (Continuous & Discrete)** and **Qualitative (Nominal & Ordinal)**.

Below we demonstrate One-Hot Encoding for Nominal data and Ordinal Mapping for Rank-ordered data.

```python
# Nominal One-Hot Encoding
nominal_data = ['Red', 'Blue', 'Green', 'Blue', 'Red']
categories = sorted(list(set(nominal_data)))
ohe_matrix = [[1 if val == c else 0 for c in categories] for val in nominal_data]

print("Categories:", categories)
for val, encoded in zip(nominal_data, ohe_matrix):
    print(f"{val:6s} -> {encoded}")

# Ordinal Encoding
ordinal_ratings = ['Low', 'High', 'Medium', 'Critical', 'Low']
mapping = {'Low': 1, 'Medium': 2, 'High': 3, 'Critical': 4}
encoded_ordinal = [mapping[r] for r in ordinal_ratings]
print(f"\nOrdinal mapped ratings: {encoded_ordinal}")
```
